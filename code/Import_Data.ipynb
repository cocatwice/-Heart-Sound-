```python
from pymongo import MongoClient
import os

# 连接到MongoDB数据库
client = MongoClient('mongodb://localhost:27017/')  
db = client['heart_sound_DB']  
collection = db['test']  

# 初始化存储结果的列表
normal_data = []
abnormal_data = []

# 定义保存音频数据的函数
def save_audio_data(audio_data, file_path):
    os.makedirs(os.path.dirname(file_path), exist_ok=True)
    with open(file_path, 'wb') as f:
        f.write(audio_data)

# 查询test集合中的所有数据
for record in collection.find():
    # 获取label1和label2的值
    label1 = record.get('label1', '').lower() if isinstance(record.get('label1', ''), str) else str(record.get('label1', ''))
    label2 = record.get('label2', '').lower() if isinstance(record.get('label2', ''), str) else str(record.get('label2', ''))

    # 获取音频数据和文件名
    audio_data = record.get('content', None)
    filename = record.get('filename', '')

    # 判断是否为normal
    if label1 == 'normal' or label2 == 'normal':
        normal_data.append(record)
        # 保存normal音频数据
        if audio_data and filename:
            save_audio_data(audio_data, f'../two_class_data/n/{filename}')
    # 判断是否为abnormal
    elif label2 == 'abnormal' or label1 != 'normal':
        abnormal_data.append(record)
        # 保存abnormal音频数据
        if audio_data and filename:
            save_audio_data(audio_data, f'../two_class_data/abn/{filename}')

print(f"Normal data count: {len(normal_data)}")
print(f"Abnormal data count: {len(abnormal_data)}")
```
