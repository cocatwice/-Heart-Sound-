```python
#kaggle数据集导入mongodb数据库
from pymongo import MongoClient
import bson.binary
from io import StringIO,BytesIO
import os
import csv
client=MongoClient('127.0.0.1',27017)


PCGdatabase=client['heart_sound_DB']
col_test=PCGdatabase['test']

for i in os.listdir('../HeartSound1/datasets/kaggle/set_a'):
    if i[0:6]=='normal':
        label1=i[0:6]
    elif i[0:8]=='extrahls':
        label1=i[0:8]
    elif i[0:6]=='murmur':
        label1=i[0:6]
    elif i[0:8]=='artifact':
        label1=i[0:8]
    else:
        label1='Aunlabelledtest'

    with open ('../HeartSound1/datasets/kaggle/set_a/'+i,'rb') as audio:
        content = BytesIO(audio.read())
        col_test.insert_one(dict(
        content= bson.binary.Binary(content.getvalue()),
        filename = i,
        label1=label1,##PASCAL中分类：normal,murmur,extrahs,artifact,Aunlabelledtest,extrasystole
        label2=False,##normal or abnormal
        posture=False,#姿势
        area=False,#听诊器放置区域
        posion=False,#听诊器听诊的位置
        period=False,#音频的时期（只针对密歇根大学）
        original_dataset='Kaggle',#原始数据集
        col_method='phone'#采集方式:stethoscope(听诊器),phone(手机)
      ))

for i in os.listdir('../HeartSound1/datasets/kaggle/set_b'):
    if i[0:6]=='normal':
        label1=i[0:6]
    elif i[0:12]=='extrasystole':
        label1=i[0:12]
    elif i[0:6]=='murmur':
        label1=i[0:6]
    else:
        label1='Bunlabelledtest'

    with open ('../HeartSound1/datasets/kaggle/set_b/'+i,'rb') as audio:
        content = BytesIO(audio.read())
        col_test.insert_one(dict(
        content= bson.binary.Binary(content.getvalue()),
        filename = i,
        label1=label1,
        label2=False,
        posture=False,#姿势
        area=False,#听诊器放置区域
        posion=False,#听诊器听诊的位置
        period=False,#音频的时期（只针对密歇根大学）
        original_dataset='Kaggle',
        col_method='stethoscope'#采集方式:stethoscope(听诊器),phone(手机)
      ))
```

```python
#将Yaseen Khan提供的五分类数据中的非normal数据导入数据库
col_four_class=PCGdatabase['four_class']

for i in os.listdir('../HeartSound1/datasets/five_class/'):
    if i[0:2] == 'AS': label = 'AS'
    elif i[0:2] == 'MR': label = 'MR'
    elif i[0:2] == 'MS': label = 'MS'
    elif i[0:3] == 'MVP': label = 'MVP'

    if i[0:1] != 'N':    
      for j in os.listdir('../HeartSound1/datasets/five_class/'+i):
          with open ('../HeartSound1/datasets/five_class/'+i+'/'+j,'rb') as audio:
            content = BytesIO(audio.read())
            col_four_class.insert_one(dict(
            content= bson.binary.Binary(content.getvalue()),
            filename = i,
            label1=label,##PASCAL中分类：normal,murmur,extrahs,artifact,Aunlabelledtest,extrasystole
            label2='abnormal',##normal or abnormal
            posture=False,#姿势
            area=False,#听诊器放置区域
            posion=False,#听诊器听诊的位置
            period=False,#音频的时期（只针对密歇根大学）
            original_dataset='five_class',#原始数据集
            col_method='stethoscope'#采集方式:stethoscope(听诊器),phone(手机)
          ))
```
